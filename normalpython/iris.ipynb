```python
import pandas as pd
```

```python
columns = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Species']
df = pd.read_csv("D:\pict\DSBDAPractical\DSBDALExam DataSets\Iris\Iris.csv", header=None, names = columns)
df.head()
```

```python
print(df.shape)
print(df.columns)

df.info()
```

```python
list(df['Species'].unique())
```

```python
setosas = df[df['Species'] == 'Iris-setosa']
display(setosas.head())

versicolors = df[df['Species'] == 'Iris-versicolor']
display(versicolors.head())
```

```python
merged_data = pd.concat([setosas, versicolors])
display(merged_data.head())
print(merged_data.shape)
```

```python
sorted_petals = df.sort_values(by = 'PetalLength') # , ascending = False
display(sorted_petals.head())
```

```python
transposed_df = df.T
display(transposed_df.head())
```

```python
melted_data = pd.melt(
    df,
    id_vars = ['Species'],
    var_name = 'Measurement',
    value_name = 'Value'
)
display(melted_data.head())
```

```python
melted_data['ID'] = melted_data.index
wide_data = melted_data.pivot_table(
    index='ID',
    columns='Measurement',
    values='Value',
    aggfunc='first'
)
display(wide_data.head())
```
