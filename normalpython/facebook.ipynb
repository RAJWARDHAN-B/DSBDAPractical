```python
import pandas as pd
```

```python
df = pd.read_csv("D:\pict\DSBDAPractical\DSBDALExam DataSets\dataset_Facebook.csv", sep=";")
df.head()
```

```python
print("Shape", df.shape)
print("\nColumns\n", df.columns)
df.info()
```

```python
photo_posts = df[df['Type'] == "Photo"]

status_posts = df[df['Type'] == 'Status']

print("Photo Posts")
display(photo_posts.head())

print("Status Posts")
display(status_posts.head())
```

```python
merged_data = pd.concat([photo_posts, status_posts])
display(merged_data.head())
print("Merged Data Shape", merged_data.shape)
```

```python
sorted_data_asc = df.sort_values(by="Page total likes")
display(sorted_data_asc.head())
```

```python
sorted_data_desc = df.sort_values(by='Page total likes', ascending=False)

display(sorted_data_desc.head())
```

```python
transposed_data = df.T
display(transposed_data.head())
```

```python
# Melt selected columns
melted_data = pd.melt(
    df,
    id_vars=['Type'],
    value_vars=['like', 'share', 'comment'],
    var_name='Interaction Type',
    value_name='Count'
)

# Display melted data
display(melted_data.head())
```

```python
# Pivot table to cast back to wide format
wide_data = melted_data.pivot_table(
    index='Type',
    columns='Interaction Type',
    values='Count',
    aggfunc='mean'
)

# Display wide format data
display(wide_data)
```
